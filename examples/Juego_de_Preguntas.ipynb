```python
# Juego de preguntas en terminal

# Esta variable controla si el juego completo se repite o no
# Es un valor booleano (True o False)
seguir_jugando = True

# Este while controla TODO el juego (permite jugar varias veces)
# Mientras la condición sea True, el programa seguirá ejecutándose
while seguir_jugando == True:

    # Pedimos datos al usuario usando input()
    # input() siempre devuelve texto (str), por eso convertimos la edad a int
    nombre = input("\n¿Cuál es tu nombre? ")
    edad = int(input("¿Cuál es tu edad? "))

    # Mostramos mensajes en pantalla
    print("\nHola", nombre, "! Bienvenido al juego.")
    print("Tienes 3 vidas. Si fallas 3 veces, pierdes.\n")

    # Variables del juego
    vidas = 3          # número de intentos antes de perder
    puntaje = 0        # respuestas correctas
    jugando = True     # controla la partida actual

    # Preguntas y respuestas (todas son variables tipo str)
    pregunta1 = "¿Capital de Francia?"
    respuesta1 = "paris"

    pregunta2 = "¿5 + 7?"
    respuesta2 = "12"

    pregunta3 = "¿Color del cielo en un día despejado?"
    respuesta3 = "azul"

    pregunta4 = "¿Animal que ladra?"
    respuesta4 = "perro"

    pregunta5 = "¿Cuántos días tiene una semana?"
    respuesta5 = "7"

    pregunta6 = "¿Planeta en el que vivimos?"
    respuesta6 = "tierra"

    pregunta7 = "¿Lenguaje de programación que estás usando?"
    respuesta7 = "python"

    pregunta8 = "¿2 * 6?"
    respuesta8 = "12"

    pregunta9 = "¿Mes con 28 días (mínimo)?"
    respuesta9 = "febrero"

    pregunta10 = "¿Agua en inglés?"
    respuesta10 = "water"

    # Esta variable nos dice en qué pregunta vamos
    numero_pregunta = 1

    # Este while controla UNA partida del juego
    # Mientras jugando sea True, el usuario sigue respondiendo
    while jugando == True:

        # Primero revisamos si el jugador ya perdió
        # Si vidas llega a 0 o menos, termina la partida
        if vidas <= 0:
            print("\nTe quedaste sin vidas. Perdiste 😢")
            jugando = False

        # ---------------- PREGUNTA 1 ----------------
        elif numero_pregunta == 1:

            # input() recibe la respuesta del usuario
            # .lower() convierte el texto a minúsculas
            # Ejemplo: "Paris", "PARIS", "paris" → todos se vuelven "paris"
            # Esto nos permite comparar sin preocuparnos por mayúsculas
            r = input(pregunta1 + " ").lower()

            # Comparamos la respuesta del usuario con la correcta
            if r == respuesta1:
                print("Correcto ✅")
                puntaje += 1   # sumamos 1 al puntaje actual
            else:
                print("Incorrecto ❌")
                vidas -= 1     # restamos 1 vida a las vidas restantes

            # Pasamos a la siguiente pregunta
            numero_pregunta += 1

        # ---------------- PREGUNTA 2 ----------------
        elif numero_pregunta == 2:
            r = input(pregunta2 + " ").lower()

            # Misma lógica que la pregunta anterior:
            # comparar respuesta, sumar puntaje o restar vidas
            if r == respuesta2:
                print("Correcto ✅")
                puntaje += 1
            else:
                print("Incorrecto ❌")
                vidas -= 1

            numero_pregunta += 1

        # ---------------- PREGUNTA 3 ----------------
        elif numero_pregunta == 3:
            r = input(pregunta3 + " ").lower()

            # Misma lógica de pregunta
            if r == respuesta3:
                print("Correcto ✅")
                puntaje += 1
            else:
                print("Incorrecto ❌")
                vidas -= 1

            numero_pregunta += 1

        # ---------------- PREGUNTA 4 ----------------
        elif numero_pregunta == 4:
            r = input(pregunta4 + " ").lower()

            # Misma lógica de pregunta
            if r == respuesta4:
                print("Correcto ✅")
                puntaje += 1
            else:
                print("Incorrecto ❌")
                vidas -= 1

            numero_pregunta += 1

        # ---------------- PREGUNTA 5 ----------------
        elif numero_pregunta == 5:
            r = input(pregunta5 + " ").lower()

            # Misma lógica de pregunta
            if r == respuesta5:
                print("Correcto ✅")
                puntaje += 1
            else:
                print("Incorrecto ❌")
                vidas -= 1

            numero_pregunta += 1

        # ---------------- PREGUNTA 6 ----------------
        elif numero_pregunta == 6:
            r = input(pregunta6 + " ").lower()

            # Misma lógica de pregunta
            if r == respuesta6:
                print("Correcto ✅")
                puntaje += 1
            else:
                print("Incorrecto ❌")
                vidas -= 1

            numero_pregunta += 1

        # ---------------- PREGUNTA 7 ----------------
        elif numero_pregunta == 7:
            r = input(pregunta7 + " ").lower()

            # Misma lógica de pregunta
            if r == respuesta7:
                print("Correcto ✅")
                puntaje += 1
            else:
                print("Incorrecto ❌")
                vidas -= 1

            numero_pregunta += 1

        # ---------------- PREGUNTA 8 ----------------
        elif numero_pregunta == 8:
            r = input(pregunta8 + " ").lower()

            # Misma lógica de pregunta
            if r == respuesta8:
                print("Correcto ✅")
                puntaje += 1
            else:
                print("Incorrecto ❌")
                vidas -= 1

            numero_pregunta += 1

        # ---------------- PREGUNTA 9 ----------------
        elif numero_pregunta == 9:
            r = input(pregunta9 + " ").lower()

            # Misma lógica de pregunta
            if r == respuesta9:
                print("Correcto ✅")
                puntaje += 1
            else:
                print("Incorrecto ❌")
                vidas -= 1

            numero_pregunta += 1

        # ---------------- PREGUNTA 10 ----------------
        elif numero_pregunta == 10:
            r = input(pregunta10 + " ").lower()

            # Misma lógica de pregunta
            if r == respuesta10:
                print("Correcto ✅")
                puntaje += 1
            else:
                print("Incorrecto ❌")
                vidas -= 1

            numero_pregunta += 1

        # Si ya no hay más preguntas, termina el juego
        else:
            print("\nJuego terminado 🎮")
            jugando = False

        # Este bloque se ejecuta después de cada pregunta
        # Solo si el juego sigue activo
        if jugando == True:
            print("Vidas:", vidas, "| Puntaje:", puntaje, "\n")

    # Cuando salimos del while de la partida, mostramos resultados finales
    print("\nResultado final:")
    print("Puntaje:", puntaje)
    print("Vidas restantes:", vidas)

    # Preguntamos si quiere volver a jugar
    respuesta = input("\n¿Quieres jugar otra vez? (si/no): ").lower()

    # Evaluamos la respuesta del usuario
    if respuesta == "si":
        seguir_jugando = True   # se repite el juego
    elif respuesta == "no":
        seguir_jugando = False  # se termina el programa
    else:
        # Si el usuario escribe algo distinto
        print("No entendí tu respuesta, se cerrará el juego.")
        seguir_jugando = False

# Cuando el while principal termina, el programa finaliza
print("\nGracias por jugar 👋")
```
