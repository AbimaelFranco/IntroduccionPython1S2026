## Operadores Aritméticos y de *Asignación*

```python
# Operadores aritmeticos
#int
#float
#bool
a = 658
b = 27

print(b + a)  # suma
print(a - b)  # resta
print(a * b)  # multiplicación
print(a / b)  # división
print(a // b) # división entera
print(a % b)  # módulo (residuo)
print(a ** b) # potencia
```

```python
# Operadores de Asignación

x = 5     # asignación
print(f"x = {x}")

x += 2    # equivalente a x = x + 2
print(f"x = {x}")

x -= 3    # equivalente a x = x + 3
print(f"x = {x}")

x *= 3    # equivalente a x = x * 3
print(f"x = {x}")
```

## Operadores lógicos
### Tipo de dato lógico (Booleano)
- `True` (verdadero / 1)
- `False` (falso / 0)

### Operaciones lógicas
- `and`: verdadero si **ambas** condiciones son verdaderas.

| A     | B     | A and B |
| ----- | ----- | ------- |
| True  | True  | True    |
| True  | False | False   |
| False | True  | False   |
| False | False | False   |

- `or`: verdadero si **al menos una** condición es verdadera.

| A     | B     | A or B |
| ----- | ----- | ------ |
| True  | True  | True   |
| True  | False | True   |
| False | True  | True   |
| False | False | False  |

- `not`: invierte el valor lógico.

| A     | not A |
| ----- | ----- |
| True  | False |
| False | True  |

```python
a = True
b = True
c = not a and b

print(c)
```

```python
# Cotizador de viaje

# precios base
precio_vuelo = 350
precio_equipaje = 40
precio_comida = 25
precio_recuerdos = 60

# decisiones de compra
equipaje_extra = True
comer_aeropuerto = False
comprar_recuerdos = True

# tipo de cambio
tipo_cambio = 7.8

total = precio_vuelo

total += precio_equipaje * equipaje_extra
total += precio_comida * comer_aeropuerto
total += precio_recuerdos * comprar_recuerdos

total_moneda_local = total * tipo_cambio

print("Resumen del viaje:")
print("Vuelo:", precio_vuelo)

print("\nEquipaje extra:", equipaje_extra)
print("Presupuesto de equipaje:", precio_equipaje)

print("\nComida en aeropuerto:", comer_aeropuerto)
print("Presupuesto de comida:", precio_comida)

print("\nRecuerdos:", comprar_recuerdos)
print("Presupuesto de recuerdos:", precio_recuerdos)

print("\nPresupuesto aproximado en dólares:", total)
print("Presupuesto en moneda local:", total_moneda_local)

print("")
print(f"El viaje tiene un costo total de ${total} equivalente aproximadamete a Q{total_moneda_local}, e incluye: \nEquipaje extra: {equipaje_extra} \nAlimentacion: {comer_aeropuerto} \nRecuerdos: {comprar_recuerdos}")
```
