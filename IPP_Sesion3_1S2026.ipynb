## Operadores lógicos
### Tipo de dato lógico
- `True` (verdadero / 1)
- `False` (falso / 0)

### Operaciones lógicas
- `and`: verdadero si **ambas** condiciones son verdaderas.

| A     | B     | A and B |
| ----- | ----- | ------- |
| True  | True  | True    |
| True  | False | False   |
| False | True  | False   |
| False | False | False   |

- `or`: verdadero si **al menos una** condición es verdadera.

| A     | B     | A or B |
| ----- | ----- | ------ |
| True  | True  | True   |
| True  | False | True   |
| False | True  | True   |
| False | False | False  |

- `not`: invierte el valor lógico.

| A     | not A |
| ----- | ----- |
| True  | False |
| False | True  |

2. Ciclos `while`

- Un ciclo **while** repite repetir condigo en base a una condición.
- Mientras se cumpla " --- " entonces " --- "
- Ejecuta el codigo mientras la condición siga siendo verdadera

```python
condicion = "SI"

while condicion == "SI":

  edad = int(input("Ingresa tu edad: "))
  print(f"La edad es: {edad}")

  condicion = input("¿Quieres continuar? (SI/NO): ")
  if condicion == "NO":
    condicion = "NO"


print("Fin del ciclo")
```

```python
# Si nota >= 61: Aprobado, caso contrario: Reprueba
condicion = 1

while condicion == 1:
  estudiante = input("Ingresa el nombre del estudiante ")
  nota = int(input(f"Ingresa la nota de {estudiante}: "))

  if nota >= 0:
    if nota >= 61:
      print(f"{estudiante} aprobó")
    else:
      print(f"{estudiante} reprobó")
  else:
    print("Nota inválida")

  condicion = int(input("Quieres ingresar más notas? "))
```
