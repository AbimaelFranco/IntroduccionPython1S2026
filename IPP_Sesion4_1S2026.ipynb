## Estructuras de datos

Las estructuras de datos son formas de organizar y almacenar información para que podamos acceder a ella y modificarla eficientemente. En Python, algunas de las más usadas son:

**Listas:** colecciones ordenadas y modificables.

**Tuplas:** colecciones ordenadas e inmutables.

**Diccionarios:** colecciones de pares clave-valor.

## Listas (`list`)

Una lista es una colección ordenada de elementos que puede cambiar (mutable).

* Se definen con corchetes [].

* Pueden almacenar distintos tipos de datos.

* Se pueden modificar, agregar o eliminar elementos.

* Se accede a los elementos con índices, empezando en 0.

```python
# Crear una lista
frutas = ["manzana", "banana", "cereza"]
print(f"Lista original: {frutas}")

# Acceder por índice
# indice = int(input("Ingrese el numero de indice: "))
# print(f"El valor en el indice {indice} es: {frutas[indice]}")

# Modificar un elemento
frutas[1] = "naranja"
print(f"Lista modificada: {frutas}")  # ['manzana', 'naranja', 'cereza']

# Agregar elementos
frutas.append("uva")
print(f"Lista con elemento extra: {frutas}")  # ['manzana', 'naranja', 'cereza', 'uva']

# Eliminar elementos
frutas.remove("naranja")
print(f"Lista con elemento eliminado: {frutas}")  # ['manzana', 'cereza', 'uva']

fruta_seleccionada = frutas[int(input("Ingrese el numero de la fruta que desea: "))]
print(f"La fruta seleccionada es: {fruta_seleccionada}")
```

## Tuplas (`tuple`)

Una tupla es una colección ordenada de elementos inmutable (no se puede cambiar después de creada).

* Se definen con paréntesis ().

* Son más rápidas que las listas.

* Útiles para datos que no deben modificarse.

```python
# Crear una tupla
coordenadas = (10, 20, 3, 4, 5, "elemento6")

# Acceder por índice
print(coordenadas[5])  # 10

# No se puede modificar
# coordenadas[0] = 15  # ❌ Error
```

## Diccionarios (`dict`)

Un diccionario almacena datos en pares clave: valor.

* Se definen con llaves {}.

* Se accede a los valores usando la clave, no el índice.

* Las claves deben ser únicas.

```python
# Crear un diccionario
persona = {
    "nombre": "Pedro",
    "edad": 25,
    "ciudad": "Madrid"
}

# Acceder a un valor por clave
print(f"El usuario {persona["nombre"]}, tiene {persona["edad"]} años y vive en {persona["ciudad"]}")  # Ana

# Modificar un valor
persona["edad"] = 26
print(f"El usuario modificado es {persona["nombre"]}, tiene {persona["edad"]} años y vive en {persona["ciudad"]}")  # Ana


# Agregar un nuevo elemento
persona["profesion"] = "Ingeniera"
print(f"El usuario {persona["nombre"]}, tiene {persona["edad"]} años, vive en {persona["ciudad"]} y su profesion es {persona["profesion"]}")  # Ana


# Eliminar una clave
del persona["ciudad"]

print(persona)
# {'nombre': 'Ana', 'edad': 26, 'profesion': 'Ingeniera'}
```

## Estructura de Control `for`

El ciclo for en Python es una estructura de control de flujo que permite iterar sobre cualquier objeto iterable.
Un iterable es cualquier objeto que puede producir sus elementos uno a uno, como:

* Listas, tuplas, diccionarios, cadenas de texto.

* Rango de números con range().

* Conjuntos (set).

* Objetos más avanzados como generadores o iteradores personalizados.

En pocas palabras:

El for ejecuta un bloque de código repetidamente para cada elemento que un iterable le proporciona.

```python
# Sintaxis:

# for variable in coleccion:
    # instrucciones
```

```python
# print(frutas)
# for i in frutas:
#   print(f"El elemento actual es: {i}")

# print(frutas)

# maximo = 5
# for i in range(len(frutas)):
#   print(f"Elemento {i+1} de la lista: {frutas[i]}")


lista_nueva = []
longitud = int(input("Ingrese la longitud de la lista: "))
for i in range(longitud):
  valor = input(f"Ingrese el elemento {i+1}: ")

  if valor == "pass":
    print("Valor no valido")
    lista_nueva.append("--")
  else:
    lista_nueva.append(valor)

print(lista_nueva)
```

```python
# Recorrer una lista
frutas = ["manzana", "banana", "cereza"]
for fruta in frutas:
    print(fruta)

print("-----------")

# Recorrer un diccionario
persona = {"nombre": "Ana", "edad": 25, "Ciudad": "Guatemala", "Estado_Civil": "Casada"}
for clave in persona:
    print(clave, ":", persona[clave])
```

## Controladores de Flujo `break`, `continue`

### `break`
Sirve para detener el ciclo antes de que termine.

```python
for numero in range(10):
    print(numero)
    if numero == 5:
        break

# Salida: 0, 1, 2, 3, 4
```

### `continue`

Sirve para saltar a la siguiente iteración sin ejecutar el resto del código en esa vuelta.

```python
for numero in range(6):
    if numero == 3:
        continue
    print(numero)
# Salida: 0, 1, 2, 4, 5
```

```python
#Codigo que muestra todos los campos (clave y valor) a excepción de la contraseña
usuario = {
    "username": "Luis199",
    "password": "98aBcD12",
    "status": True,
    "name": "Luis",
    "Lastname": "Perez",
    "age": 21,
    "country": "Guatemala"
}

for key in usuario:
  if key == "password":
    continue
  print(key, ":", usuario[key])

print("---------")

for key, value in usuario.items():
    if key == "password":
        continue
    print(key, ":", value)
```

```python
#Codigo que solicita ingresar un valor para cada campo
usuario = {
    "username": "",
    "status": False,
    "name": "",
    "Lastname": "",
    "age": 0,
    "country": ""
}

for key in usuario:
  valor = input(f"Ingrese el valor de {key}: ")
  usuario[key] = valor

print(usuario)
```

```python
#Codigo para verificacion de contraseña
usuario["password"] = "asdf1234"

while True:
  psw = input("Ingrese la contraseña: ")
  if psw == usuario["password"]:
    print(usuario)
    break
  else:
    print("Contraseña incorrecta")
```
